```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
dataset = pd.read_csv("../data/Algerian.csv")

dataset.head()
```

```python
dataset.info()
```

```python
# data cleaning
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122,"Region"]=1
df=dataset
```

```python
df.info()
```

```python
df[['Region']] = df[['Region']].astype('Int64')
df.head()
```

```python

df.isnull().sum()
```

```python
# removing the null values
df=df.dropna().reset_index(drop=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
objects = [col for col in df.columns if df[col].dtype == 'object']
```

```python
# Changing the other columns to float datatypes
for col in objects:
    if col != 'Classes':
        df[col] = df[col].astype(float)
```
