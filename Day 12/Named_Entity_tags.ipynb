```python
# Named Entity Recoginition
sentence="The Eiffel Tower is located in Paris, France. It is one of the most famous landmarks in the world."
```

```python
sentence
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python


# Download tokenizer + POS tagger data if missing
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')

# # Now tag the words
# tagged_words = nltk.pos_tag(words)
# print(tagged_words)
```

```python
nltk.download('maxent_ne_chunker')
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker_tab')

named_entities = nltk.ne_chunk(tag_elements)
print(named_entities).draw()
```
