```python
# STEMMING
"""Stemming is the process of reducing a word to its base or root form. It is a common technique used in natural language processing (NLP) and information retrieval to improve the efficiency and accuracy of text analysis.
The main goal of stemming is to group together different forms of a word so that they can be treated as a single entity. For example, the words "running," "runner," and "ran" can all be reduced to the stem "run." This allows for better matching of related terms in search queries and text analysis.
There are several stemming algorithms available, with the most popular being the Porter Stemmer, Snowball Stemmer, and Lancaster Stemmer. Each algorithm has its own set of rules for reducing words to their stems, and they may produce different results for the same input word.
    """
```

```python
words=["running", "runner", "ran", "easily", "fairly"]
```

```python
#  Porter Stemmer
from nltk.stem import PorterStemmer
```

```python
stremmer = PorterStemmer()
for word in words:
    print(word+"---->"+stremmer.stem(word))
```

```python
stremmer.stem("Congratulation")
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stremmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
for word in words:
    print(word+"---->"+reg_stremmer.stem(word))
```

```python
reg_stremmer.stem('ingeating')
```
