```python
corpus="""This is a simple corpus for tokenization. It contains multiple sentences, punctuation, and various cases. The goal is to test the tokenization process effectively!\n"""
```

```python
print(corpus)
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize


```

```python
import nltk

nltk.download("punkt_tab")
sent_tokenize(corpus)
```

```python
document = sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
#  Tokenixation
# paragraph -->words
#  sentecne -->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize

for sentence in document:
    print(wordpunct_tokenize(sentence))
```
